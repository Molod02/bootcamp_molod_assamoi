```python
def my_func(*args):
    print(args)

def int_check(g,*args):
    if type(args[-1])!=int:
        print("error")
    else:
        g(*args)
        
int_check(my_func, 1,2,3)
```

```python
def my_func(*args):
    print(args)

def y_func(g):
    def int_internal_check(*args):
        if type(args[-1])!=int:
            print("error")
        else:
            g(*args)
    return int_internal_check
        
h = y_func(my_func)
h(1,2,3)
```

```python
@y_func
def my_func(*args):
    print(args)
    
my_func(1,2,'a')
```

```python

def my_debug(f):
    def new_func(*arg):
        print("arg length: "+str(len(arg)))
        f(*arg)
        # print(arg)
    return new_func
    
# a_func = my_debug( my_func )
# a_func(1,2,3)

@my_debug
def my_func(*args):
    print(args)

my_func(1,2,3)
```

```python
def log_call(func):
    def wrapper(*args, **kwargs):
        print(f"Function {func.__name__} called at {datetime.now()}")
        return func(*args, **kwargs)
    return wrapper
```

```python
@log_call
def calc_mean_std_logged(lst):
    return calc_mean_std(lst)
```

```python
calc_mean_std_logged([1, 2, 3, 4, 5])
```
